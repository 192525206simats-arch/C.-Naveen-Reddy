```python
# Candidate Elimination Algorithm

data = [
    ["Big", "Red", "Circle", "No"],
    ["Small", "Red", "Triangle", "No"],
    ["Small", "Red", "Circle", "Yes"],
    ["Big", "Blue", "Circle", "No"],
    ["Small", "Blue", "Circle", "Yes"]
]

attributes = ["Size", "Color", "Shape"]

S = ["Ø", "Ø", "Ø"]
G = [["?", "?", "?"]]

def covers(h, x):
    for i in range(len(h)):
        if h[i] != "?" and h[i] != x[i]:
            return False
    return True

for row in data:
    x = row[:3]
    target = row[3]

    if target == "Yes":

        G = [g for g in G if covers(g, x)]

        if S[0] == "Ø":
            S = x.copy()
        else:
            for i in range(3):
                if S[i] != x[i]:
                    S[i] = "?"

    else:

        new_G = []

        for g in G:
            if covers(g, x):
                for i in range(3):
                    if g[i] == "?":
                        for value in ["Big", "Small"] if i == 0 else \
                                     (["Red", "Blue"] if i == 1 else
                                      ["Circle", "Triangle"]):

                            if value != x[i]:
                                new_h = g.copy()
                                new_h[i] = value
                                new_G.append(new_h)
            else:
                new_G.append(g)

        G = new_G

# Remove duplicate hypotheses
G = [list(x) for x in set(tuple(g) for g in G)]

print("Specific Boundary (S):")
print(S)

print("\nGeneral Boundary (G):")
for g in G:
    print(g)

print("\nAll consistent hypotheses are between S and G.")
```

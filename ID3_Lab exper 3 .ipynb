```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, export_text
from sklearn.preprocessing import LabelEncoder

data = pd.DataFrame({
    'Weather': ['Sunny','Sunny','Rainy','Rainy','Overcast',
                'Sunny','Rainy','Overcast'],
    'Wind': ['Weak','Strong','Weak','Strong','Weak',
             'Weak','Weak','Strong'],
    'Play': ['No','No','Yes','No','Yes','Yes','Yes','Yes']
})

encoders = {}
for col in data.columns:
    encoders[col] = LabelEncoder()
    data[col] = encoders[col].fit_transform(data[col])

X = data[['Weather', 'Wind']]
y = data['Play']

model = DecisionTreeClassifier(criterion='entropy', random_state=0)
model.fit(X, y)

print("Decision Tree:")
print(export_text(model, feature_names=['Weather', 'Wind']))

sample = pd.DataFrame([{
    'Weather': encoders['Weather'].transform(['Overcast'])[0],
    'Wind': encoders['Wind'].transform(['Weak'])[0]
}])

result = model.predict(sample)[0]
print("New Sample: Weather=Overcast, Wind=Weak")
print("Prediction:", encoders['Play'].inverse_transform([result])[0])
```
